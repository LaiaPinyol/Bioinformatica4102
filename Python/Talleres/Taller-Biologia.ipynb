#### 7.- Con la secuencia y la lista siguiente, escribe un algoritmo que cuente el número de veces que aparece un elemento de la lista en la secuencia. Pista: hacer un diccionario para almacenar los resultados puede ser buena idea

```python
dna = "AATGATGAACGAC" 
dinucleotides = ['AA','AT','AG','AC', 
                 'TA','TT','TG','TC', 
                 'GA','GT','GG','GC', 
                 'CA','CT','CG','CT'] 
```

```python
diccionario_dna = {}
for dincl in dinucleotides:
    diccionario_dna[dincl] = dna.count(dincl)

diccionario_dna
```

##### ¿Cuántos counts tiene TC?

```python
diccionario_dna["TC"]
```

##### ¿Puedes imprimir los dinucleótidos que tengan counts igual a 2?

```python
for dincl, count in diccionario_dna.items():
    if count == 2:
        print(dincl)
```
